```python

import pandas as pd
from kafka import KafkaConsumer
from time import sleep
from json import dumps,loads
import json
from s3fs import S3FileSystem
```

```python
consumer = KafkaConsumer('demo_testing',
                         bootstrap_servers=['51.20.70.146:9092'],
                         value_deserializer=lambda x: loads(x.decode('utf-8')))
```

```python
# for c in consumer:
#     print(c.value)
```

```python
for count,i in enumerate(consumer):
    print(count)
    print(i.value)
```

```python
s3 = S3FileSystem(client_kwargs={'region_name': 'eu-north-1'})

for count, i in enumerate(consumer):
    with s3.open("s3://kafka-stock-market-project-manisha/stock_market_{}.json".format(count), 'w') as file:
        json.dump(i.value, file)
```
